## Configuração do Ambiente

Execute a célula abaixo antes dos exemplos: importa as bibliotecas, carrega o `vendas.csv` (mesmo arquivo dos tutoriais de pandas e matplotlib) e aplica o tema padrão do seaborn.

```python
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# DataFrame de vendas (mesmo arquivo dos tutoriais anteriores)
df = pd.read_csv('vendas.csv', sep=';', encoding='utf-8')
df['data_hora'] = pd.to_datetime(df['data_hora'], format='%Y-%m-%d %H:%M:%S')

df['mes'] = df['data_hora'].dt.month_name().str[:3]
df['periodo'] = np.where(df['data_hora'].dt.month <= 3, 'Trimestre 1', 'Trimestre 2')

# Tema padrão do seaborn (grade suave, fundo branco)
sns.set_theme()

print('Dados carregados:', df.shape)
```

# MÓDULO 1 — Nível Básico
## O Essencial: Distribuições, Dados Categóricos e Relações

---
## 1.1 Distribuições de Dados

### `sns.histplot()` — Histograma

Distribui uma variável contínua em intervalos, mostrando a **frequência** dos valores. Aceita `hue` para separar por categoria.

**Sintaxe:** `sns.histplot(data=df, x='coluna', bins=n, kde=True)`

```python
sns.histplot(data=df, x='valor', bins=10, kde=True, color='steelblue')
plt.title('Distribuição dos Valores')
plt.show()
```

### `sns.kdeplot()` — Curva de Densidade

Estimativa suavizada da distribuição. Perfeita para **comparar grupos** usando o parâmetro `hue`.

**Sintaxe:** `sns.kdeplot(data=df, x='coluna', hue='grupo', fill=True)`

```python
sns.kdeplot(data=df, x='valor', hue='regiao', fill=True, alpha=0.4)
plt.title('Densidade dos Valores por Região')
plt.show()
```

### `sns.ecdfplot()` — Função de Distribuição Acumulada

Mostra a **proporção acumulada** dos dados: qual percentual está abaixo de cada valor.

**Sintaxe:** `sns.ecdfplot(data=df, x='coluna')`

```python
sns.ecdfplot(data=df, x='valor', hue='regiao')
plt.title('Curvas ECDF dos Valores por Região')
plt.show()
```

---
## 1.2 Dados Categóricos

### `sns.countplot()` — Contagem por Categoria

Conta o número de ocorrências de cada categoria. É o equivalente gráfico do `value_counts()`.

**Sintaxe:** `sns.countplot(data=df, x='coluna')`

```python
sns.countplot(data=df, x='estado')
plt.title('Vendas por Estado')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### `sns.barplot()` — Média com Incerteza

Mostra a **média** de uma variável por categoria, com **barras de erro** (intervalo de confiança) por padrão.

**Sintaxe:** `sns.barplot(data=df, x='categoria', y='valor')`

```python
sns.barplot(data=df, x='categoria', y='valor', hue='categoria', palette='muted', legend=False)
plt.title('Valor Médio por Categoria')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### `sns.boxplot()` — Diagrama de Caixa

Resume a distribuição com **quartis, mediana e outliers**, ideal para comparar grupos.

**Sintaxe:** `sns.boxplot(data=df, x='categoria', y='valor')`

```python
sns.boxplot(data=df, x='categoria', y='valor')
plt.title('Distribuição dos Valores por Categoria')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### `sns.violinplot()` — Gráfico de Violino

Combina boxplot com **curva de densidade** para mostrar a forma completa da distribuição.

**Sintaxe:** `sns.violinplot(data=df, x='categoria', y='valor', split=...)`

```python
sns.violinplot(data=df, x='categoria', y='valor', hue='categoria', palette='pastel', legend=False)
plt.title('Distribuição (Violino) por Categoria')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### `sns.boxenplot()` — Boxplot em Letras

Variação do boxplot que mostra **muito mais detalhes** da cauda da distribuição (ótima para dados grandes).

**Sintaxe:** `sns.boxenplot(data=df, x='categoria', y='valor')`

```python
sns.boxenplot(data=df, x='categoria', y='valor')
plt.title('Distribuição (Boxen) por Categoria')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

---
## 1.3 Relações Entre Variáveis

### `sns.scatterplot()` — Dispersão

Plota a relação entre duas variáveis. Use `hue`, `size` e `style` para **adicionar dimensões extras**.

**Sintaxe:** `sns.scatterplot(data=df, x='col1', y='col2', hue='grupo')`

```python
sns.scatterplot(data=df, x='valor', y='valor_venda', hue='regiao', style='status', s=80)
plt.title('Valor x Valor de Venda — Relações')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### `sns.lineplot()` — Linha com Intervalo de Confiança

Agrega dados ao longo do eixo X mostrando a média e a **faixa de incerteza** por padrão.

**Sintaxe:** `sns.lineplot(data=df, x='data', y='valor', marker='o')`

```python
serie = df.groupby('mes')['valor'].sum().reset_index()

sns.lineplot(data=serie, x='mes', y='valor', marker='o', linewidth=2)
plt.title('Vendas (Soma) por Mês')
plt.xlabel('Mês')
plt.ylabel('Valor (R$)')
plt.show()
```

### `sns.jointplot()` — Dispersão + Marginais

Mostra a dispersão central com a **distribuição marginal** de cada variável nas bordas.

**Sintaxe:** `sns.jointplot(data=df, x='col1', y='col2', kind='scatter')` — kinds: `scatter`, `kde`, `hex`, `reg`

```python
sns.jointplot(data=df, x='valor', y='valor_venda', kind='scatter', hue='status')
plt.show()
```

# MÓDULO 2 — Nível Intermediário
## Temas, Paletas, Grids e Heatmaps

---
## 2.1 Temas e Paletas

### `sns.set_theme()` — Estilos Prontos

Configura o estilo global. O parâmetro `style` controla fundo e grade: `darkgrid`, `whitegrid`, `dark`, `white`, `ticks`.

**Sintaxe:** `sns.set_theme(style='whitegrid', context='notebook')` — contexts: `paper`, `notebook`, `talk`, `poster`

```python
sns.set_theme(style='whitegrid')

sns.countplot(data=df, x='status', hue='regiao')
plt.title('Vendas por Status e Região — Estilo Whitegrid')
plt.show()
```

### Paletas de Cores

Dá para controlar as cores com `palette` (nomes conhecidos) ou criar paletas com `sns.color_palette()`.

**Sintaxe:** `sns.barplot(..., palette='pastel')` e `sns.color_palette('viridis', n_colors=5)`

```python
# Meses abreviados em inglês (padrão do month_name())
meses_foco = ['Jan', 'Feb', 'Mar', 'Apr']
dados_foco = df[df['mes'].isin(meses_foco)]
paleta_viridis = sns.color_palette('viridis', n_colors=dados_foco['mes'].nunique())

sns.barplot(data=dados_foco, x='mes', y='valor',
            hue='mes', palette=paleta_viridis, legend=False)
plt.title('Valores por Mês — Paleta Viridis')
plt.show()
```

---
## 2.2 Grids de Facetas: `relplot`, `catplot`, `displot`

### `sns.relplot()` — Dispersão/Linha com Facetas

Cria uma grade de subplots (facetas) combinando `hue`/`col`/`row`. É a versão *Grid* do scatterplot e lineplot.

**Sintaxe:** `sns.relplot(data=df, x='..', y='..', col='categoria', kind='scatter')` — kinds: `scatter`, `line`

```python
g = sns.relplot(data=df, x='valor', y='valor_venda', col='regiao',
                hue='status', s=80, aspect=1.1)
g.figure.suptitle('Dispersão por Região', y=1.05)
plt.show()
```

### `sns.catplot()` — Dados Categóricos com Facetas

Versão *Grid* para os gráficos categóricos. O `kind` define o tipo: `box`, `violin`, `bar`, `count`, `boxen`, `strip`, `point`.

**Sintaxe:** `sns.catplot(data=df, x='..', y='..', col='..', kind='box')`

```python
sns.catplot(data=df, x='categoria', y='valor', col='regiao',
            kind='box', hue='periodo')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### `sns.displot()` — Distribuições com Facetas

Versão *Grid* dos gráficos de distribuição (`hist`, `kde`, `ecdf`).

**Sintaxe:** `sns.displot(data=df, x='coluna', col='grupo', kind='hist', kde=True)`

```python
sns.displot(data=df, x='valor', col='status', kind='hist', kde=True, bins=8)
plt.show()
```

---
## 2.3 Heatmaps

### `sns.heatmap()` — Tabela de Cores

Transforma uma matriz (pivot ou correlação) em uma grade colorida. `annot=True` exibe os valores.

**Sintaxe:** `sns.heatmap(data=pivot, annot=True, cmap='vlag', fmt='.0f')`

```python
pivot = pd.pivot_table(df, values='valor', index='mes', columns='estado', aggfunc='sum', fill_value=0)

plt.figure(figsize=(9, 5))
sns.heatmap(pivot, annot=True, fmt='.0f', cmap='YlGnBu', linewidths=0.5)
plt.title('Heatmap: Valor por Mês x Estado')
plt.tight_layout()
plt.show()
```

### Heatmap de Correlação

Aplicação clássica: visualizar a **correlação** entre variáveis numéricas do DataFrame.

**Sintaxe:** `sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', vmin=-1, vmax=1)`

```python
corr = df[['valor', 'valor_venda', 'cliente_id', 'temp']].corr()

plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, square=True)
plt.title('Matriz de Correlação')
plt.tight_layout()
plt.show()
```

# MÓDULO 3 — Nível Avançado
## Regressão, Facetas Complexas e Estruturas Especiais

---
## 3.1 Modelos de Regressão

### `sns.regplot()` — Dispersão + Regressão

Adiciona a **reta de regressão** e a faixa de confiança ao gráfico de dispersão.

**Sintaxe:** `sns.regplot(data=df, x='col1', y='col2', ci=95)`

```python
sns.regplot(data=df, x='valor', y='valor_venda', ci=95, scatter_kws={'alpha': 0.6})
plt.title('Regressão: Valor x Valor de Venda')
plt.show()
```

### `sns.lmplot()` — Regressão com Facetas

Executa uma regressão **por grupo/faceta**, permitindo comparar as relações entre categorias.

**Sintaxe:** `sns.lmplot(data=df, x='..', y='..', hue='grupo', col='..')`

```python
sns.lmplot(data=df, x='valor', y='valor_venda', hue='regiao', ci=None)
plt.show()
```

---
## 3.2 Comparação de Muitas Variáveis

### `sns.pairplot()` — Matriz de Dispersão

Plota **todas as combinações** entre variáveis numéricas de uma vez, com distribuições na diagonal. O `hue` coloriza os grupos.

**Sintaxe:** `sns.pairplot(data, vars=[...], hue='grupo')`

```python
sns.pairplot(data=df, vars=['valor', 'valor_venda', 'temp'], hue='status', diag_kind='kde', height=2.5)
plt.show()
```

### `sns.clustermap()` — Heatmap Hierárquico

Heatmap que **reordena linhas e colunas** por agrupamento hierárquico, revelando padrões.

**Sintaxe:** `sns.clustermap(data, cmap='vlag', annot=True, fmt='.0f')`

```python
pivot_cat = pd.pivot_table(df, values='valor', index='categoria',
                           columns='estado', aggfunc='sum', fill_value=0)

g = sns.clustermap(pivot_cat, cmap='vlag', annot=True, fmt='.0f',
                   linewidths=0.5, figsize=(8, 6))
g.fig.suptitle('Hierarquia: Categoria x Estado', y=1.05)
plt.show()
```

---
## 3.3 Plotagem em Eixos Existentes (API Orientada a Objetos)

Todos os `sns.*plot` podem receber um objeto `ax` existente via `ax=...`, permitindo **combinar seaborn e matplotlib** no mesmo gráfico.

```python
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(12, 4))

sns.barplot(data=df, x='categoria', y='valor', hue='categoria', palette='rocket', ax=ax1, legend=False)
ax1.set_title('Barplot (seaborn)')
ax1.tick_params(axis='x', rotation=45)

sns.lineplot(data=df.groupby('mes')['valor'].sum().reset_index(),
             x='mes', y='valor', marker='o', ax=ax2, color='coral')
ax2.set_title('Lineplot (seaborn)')

plt.tight_layout()
plt.show()
```
